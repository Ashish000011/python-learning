```python
# ANSWER NO - 4.
# Write a recursive function to 
# calculate the sum of first `n` natural numbers.

def sum_n(n):
    if (n==1):
        return 1 
    
    elif (n == 0):
        return 0
    
    else:
        return (n+ sum_n(n-1))

n = int(input("Enter the number of first natural numbers sum you want :"))

print(f"the sum of first {n} numbers is : {sum_n(n)}")
```

```python
# ANSWER NO - 6.
# 6.Write a Python function which converts inches to centimeters.

def inch_to_cm(n):
    return(n*2.54)

n = float(input("Enter the inches you want to know in centimeter :"))

print(f"{n} inches have {inch_to_cm(n):.2f} cm")
```

```python
# ANSWER NO - 7.

# 7. Write a Python function to remove a given word
# from a list and strip it at the same time.


def clean_list(my_list , word):
    new_list =[]  # avoid modifying original list while iterating
        
    for item in my_list:
        item = item.strip()  # it will remove spaces and all
        
        if item != word:     # if not the word we wanna remove
            new_list.append(item)
    return new_list

my_list = []

n = int(input("enter how many iteams you want in the list :"))

for i in range(1 , n+1):
    a=input("enter the word you want in the list :")
    
    my_list.append(a)
    
print("Orignal list :", my_list)

word = input("enter the word you wanna remove: ")

result = clean_list(my_list , word)

print("Item removed list is here: ", result)
    
    
```
