## Python String Slicing Quick Notes

**Rule:**

```
string[start : end]
```

* `start` index → **included**
* `end` index → **excluded**

Example string:

```
"ashish"
 a  s  h  i  s  h
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

**Examples**

```
shri[0:5]   → "ashis"
```

Indexes used:

```
0 1 2 3 4
a s h i s
```

```
shri[-4:-1] → "his"
```

Indexes used:

```
-4 -3 -2
 h  i  s
```

**Key Idea**

```
start ≤ index < end
```

Works the **same for positive and negative indexing**.

## Escape Sequence Characters

Sequence of characters after backlash"\" - <u>Escape sequence characters
    
Escape sequence characters comprise of more than one character but represents one character when used within the strings. 


     Example -   \n , \t ,  \;     \\   etc
                
                  ^    ^    ^       ^
                  |    |    |       |
            newline   tab  single  backslash
                           quote

````python
a = "ashish is a good boy \nbut not a bad boy \t hello ji \\ hehe  \"lmao\""

print(a)

# used to to string operations and all
````
