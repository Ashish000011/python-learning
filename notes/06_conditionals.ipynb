# IF ELSE AND ELIF IN PYTHON

If else and Elif statements are a multiway decision taken by our program due to certain conditions in our code.

syntax:

```python
if(condition1):       # if condition 1 is true 
    print("yes")

elif(condition2):     #if condition 2 is true
    print("no")

else:                 #otherwise
    print("maybe")
```

CODE EXAMPLE:

```python
a=22

if(a>9):
    print("greater")
else:
    print("lesser")
    
    
# the space between if and print is called -> INDENTATION SPACE
```

```python
a = int(input("enter your age : "))

if(a>=18):
    print("eligible for vote in india")
    print("good for you , lol ")
else:
    print("Not eligible for vote in india")
    print("Not bad just wait to be 18")
    
```

```python
# ANSWER N0 - 1. 

print ("find greatest between four numbers")

a = int(input("Enter first no  :"))
b = int(input("Enter second n0 :"))
c = int(input("Enter third no  :"))
d = int(input("Enter fourth no :"))

if(a>b and a>c and a>d):
    print("a is greatest :",a )

elif(b>a and b>c and b>d):
    print("b is greatest :",b)

elif(c>a and c>b and c>d):
    print("c is greatest :",c)

else:
    print("d is greatest :",d)
    
   
```

```python
# ANSWER NO - 2.
# Write a program to find out whether a student is passed or fails,**  
#    - Passing requires a total of 40% and at least 33% in each subject.  
#    - Assume 3 subjects and take marks as input from the user.

a = int(input("Enter 1st subject obtained marks :"))
a1 = int(input("Enter 1st subject whole marks :"))

b = int(input("Enter 2nd subject obtained marks :"))
b1 = int(input("Enter 2nd subject whole marks :"))

c = int(input("Enter 3rd subject obtained marks :"))
c1 = int(input("Enter 3rd subject whole marks :"))

total_obtained_marks = a+b+c
total_maximum_marks = a1+b1+c1

total_obtained_percentage = (total_obtained_marks/total_maximum_marks)*100

cond1 = a>= (33*a1)/100
cond2 = b>= (33*b1)/100
cond3 = c>= (33*c1)/100

if(total_obtained_percentage <40):
    print("Failed: overall percentage is below 40%")

elif not(cond1 and cond2 and cond3):
    print("Failed due to subjects ")
    
    if not (cond1):
        print("- subject 1")
        
    if not (cond2):
        print("- subject 2")
        
    if not (cond3):
        print("- subject 3")
    
else:
    print("Passed!! total marks:", total_obtained_marks)
    
print("percentage you got => ", total_obtained_percentage)
```

```python
# ANSWER NO - 3.
# A spam comment is defined as a text containing the following keywords:**  
#    - "Make a lot of money"  
#    - "buy now"  
#    - "subscribe this"  
#    - "click this"  
#    Write a program to detect these spams.


a = input("enter the comment or lines :").lower()

if("make a lot of money" in a):
    print("spam detected")

elif("buy now" in a):
    print("spam detected")

elif("subscribe this" in a):
    print("spam detected")

elif("click this" in a):
    print("possibly a spam")
    
else:
    print("no spam detected")
```

```python
# ANSWER NO - 6.
# Write a program to calculate the grade of a student
# from his marks using the following scheme:
#    - 90 – 100 → Ex  
#    - 80 – 90 → A  
#    - 70 – 80 → B  
#    - 60 – 70 → C  
#    - 50 – 60 → D  
#    - <50 → F  

a = int(input("Enter your marks to get to know grade :"))

if (a<=100 and a>=90):
    print("Your grade is - Ex")
    
elif(a<=90 and a>=80):
    print("Your grade is - A")

elif(a<=80 and a>=70):
    print("Your grade is - B")

elif(a<=70 and a>=60):
    print("Your grade is - C")

elif(a<=60 and a>=50):
    print("Your grade is - D")

elif (a<0):
    print("invalid marks")

else:
    print("Your grade is - F")
```
