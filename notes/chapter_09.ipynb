### READING A FILE IN PYTHON

```python
# Open the file in read mode
f = open("this.txt","r")

# read its contents
text = f.read()

# print its contents
print(text)

#close the file
f.close
```

## WRITE FILES IN PYTHON

In order to write a file , we first open it in write or append mode after which , we use the python's f.write() meathod to write the file!

```python
# open the file in write mode

f = open("meow.txt","w")
#write a string to the file
f.write("this is new shit im writting here")
#close the file 
f.close()
```

## WITH STATEMENT

The best way to open and close the file automatically is the with statement:

```python
# open the file in reaf mode using 'with' , 
# which is automatically closes the file

with open("meow.txt","r") as f:
    #read the contents of the file
    text = f.read()
    
#print the contents
print(text)
```

```python
# ASNWER NO - 3.

# Write a program to generate multiplication tables 
# from 2 to 20 and write them to different files.

# Place these files in a folder for a 13-year-old.

def generateTable(n):
    table = ""
    for i in range(1,11):
        table += f"{n} x {i} ={n*i}\n"
        
    with open(f"tables/table_{n}.txt","w") as f:
        f.write(table)
        
        
for i in range(2,21):
    generateTable(i)
```

```python
# ANSWER NO - 5.
# Repeat program 4 for a list of such words to be censored

# OKEY SO I HAVE A PARAGRAPH WITH WORDS LIKE DONKEY 
# MONKEY AND PIG AND ILL CENSORE THEM ALL BY USING A LIST 
# AND A FOR LOOP

bad_words = ["monkey","donkey","pig"]

with open("question5_fileio.txt","r")as f:
    data = f.read()
    
for word in bad_words:
    data = data.replace(word,"*" * len(word))
    
with open("question5_fileio.txt","w")as f:
    f.write(data)
```
