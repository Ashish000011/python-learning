### List Indexing

A list can be indexed just like a string.

```python
L1 =[7,9,"ashish"]

print(L1[0])  #prints 7
print(L1[1])  #prints 9
print(L1[0:2]) #prints [7,9]   --> it's list slicing
print(L1[70]) #prints error
```
