# TYPES OF LOOPS IN PYTHON

Primarily there are two types of loops in python.
> while loops

> for loops

we will look into these one by one.

### WHILE LOOPS

Syntax:

````python
while (condition):   #the block keeps executing until the condition is true 
    # body of the loop
````

In while loops, condition is cheaked first. if it evalutes to true, the body of the loops is executed otherwise not!

if the look is entered, the process of [condition cheak & execution] is continued until the condition becomes False.

````python
i = 1 
 
while(i<6):
    print(i)
    i += 1


# output: 
# 1
# 2
# 3
# 4
# 5
# eroor loop over 
````

## FOR LOOP WITH ELSE

An optional else can be used with a for loop if the code is to be executed when the loops exhausts.

##### Example:


````python
l = [1,7,8]

for item in l:
    print(item)
else:
    print("done")  #this is printed when the loop exhausts!
````

## THE BREAK AND CONTINUE STATEMENTS:

BREAK = 'Break' is used to come out of the loop when encountered. it instructs the program to exit the loop now.

CONTINUE = 'Continue is used to instruct the program to skip the   particular iteration its encountering'

##### Example:

````python
# Break and continue

for i in range (100):
    if (i == 34):
        break    # exit the loop right now 
    print(i)
    

for i in range (100):
    if (i == 34):
        continue  # skip this iteration 
    print(i)
````

## PASS STATEMENT :

Pass is a null statement in python.

it instructs to 'do nothing'.

##### Example:

````python
l = [1,7,8]

for item in l:
    pass      # without pass program would return error
````

````python
for i in range (645):
    pass 

i = 0

while (i<45):
    print(i)
    i += 1 
````

````python
# ANSWER NO - 1.
# Write a program to print multiplication table 
# of a given number using `for` loop.


# SO WE NEED TO TAKE INPUT FIRST

j = int(input("Enter the number you want to print table of :"))

for i in range (1, 11) :
    print(f"{j} X {i} = {j * i }")
    
    
````

````python
# ANSWER NO - 2.

# Write a program to greet all the person
# names stored in a list `l` and which start with 'S'. 
   
#    l = ["Harry", "Soham", "Sachin", "Rahul"]





# List of names (this is a LIST, not a string)
names = ["Harry", "Soham", "Sachin", "Rahul"]

# Loop through each item in the list
for name in names:
    # 'name' is NOT the whole list
    # 'name' is ONE string at a time (e.g., "Harry", "Soham")

    # startswith() works on STRINGS, not on LIST
    if name.startswith("S"):
        # If the name starts with 'S', greet that person
        print("Hello", name)

# Important concept:
# names → list ❌ (no startswith)
# name  → string ✅ (has startswith)
````

````python
# ANSWER NO - 3.
# Attempt problem 1 using while loop.

# SO I HAVE TO WRITE A PROGRAM WHICH WILL RETURN THE TABLE OF 
# GIVEN NO BY WHILE LOOP NOW

num = int(input("Enter the number you want to know the table of :"))

i = 1
while (i<= 10):
    print(f"{num} x {i} = {num*i}")
    i = i +1
````

````python
# ANSWER NO - 4.
# Write a program to find whether a given number is prime or not.

# LETS TAKE THE NUMBER INPUT 

j = int(input("Enter the number you want to cheak if that's prime or not :"))

for i in range (2 , j):
    if(j%i) == 0 :
        print("number is not prime")
        break
else:
    print("number is prime")
````

````python
# ANSWER NO - 5.

# Write a program to find the sum of first 
# n natural numbers using while loop.

# SO THE FORMULA OF SUM OF FIRST N NATURAL NUMBER IS = N(N+1)/2
# ALL I NEED TO DO IS IMPLEMENT IT HERE WHILE TAKING N FROM USER

n = int(input("Enter the number How many first natural number sum you want : "))

j = 1

total = 0

while j <= n:
    total = total + j
    j = j + 1
    
print("sum is :", total)
  

````

````python
# ANSWER NO - 6.

# Write a program to calculate the factorial
# of a given number using for loop.

n = int(input("Enter the number you want to find factorial of :"))

fact = 1

for i in range(1, n+1):
    fact = fact*i 
    
print("Factorial is :", fact)
    
````
