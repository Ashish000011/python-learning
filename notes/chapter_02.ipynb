## Operators in python

1. Arthematic Operators : +,-,*,/ rtc..
2. Assignment Operators : = , +=,-= etc..
3. Comparision Operators : == ,> ,< ,<= ,>= ,!= , Etc..
4. Logical Operators : AND ,OR , NOT

```python
# ARTIHEMATIC OPERATOR

A = 7
B = 4
C = A+B 

print (C)



# ASSIGNMENT OPERATORS

A = 4-2 # ASSIGN 4-2 IN A 
print (A)
B = 6 
# B += 3 # INCREMENT THE VALUE OF B BY 3 AND THEN ASSIGN IT TO B 
B -= 3 #DECREMENT THE VALUE OF B BY 3 AND THEN ASSIGN IT TO B 

print(B)



#COMPARIOSION OPERATORS - THEY ALWAYS HAVE A RETURN OF BOOLEAN

D = 5>4
print(D)
E = 5<4
print(E)
F = 5!= 5
print(F)


# Logical OPERATORS 


G = True or False
print(G)
 
H = True and False
print(H)

print(not(False))
```

### TRUTH TABLES BY PROGRAM [OBV. I MADE IT USING AI TO SAVE TIME :)  ]


```python
# Truth Table for Logical Operators

values = [True, False]

print("TRUTH TABLE FOR AND (A and B)")
print("A\tB\tA and B")
for A in values:
    for B in values:
        print(A, "\t", B, "\t", A and B)

print("\nTRUTH TABLE FOR OR (A or B)")
print("A\tB\tA or B")
for A in values:
    for B in values:
        print(A, "\t", B, "\t", A or B)

print("\nTRUTH TABLE FOR NOT (not A)")
print("A\tnot A")
for A in values:
    print(A, "\t", not A)
```
