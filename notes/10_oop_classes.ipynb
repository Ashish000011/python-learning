## SELF PARAMETER

Self refers to the instance of the class. It is automatically passed with a function call from an object.

```python
# self parameter 
 

class Employee:
    language = "py"  #this is a class attribute
    salary = 100000
    
    def getInfo(self):
        print(f"The language is {self.language}  The salary is {self.salary}")
        
    def greet(greet):
        print("good morning")
ashish = Employee()
ashish.name = "Ashish" 
ashish.language = "javaScript" #this is a object/instance attribute
# print(ashish.salary , ashish.language, ashish.name)

ashish.greet()
ashish.getInfo()

```

## STATIC METHOD

Sometimes we need a function that does not use the self- parameter. we can define a static meathod like this:

```python
@staticmethod # decorator to mark greet as a static method
def greet():
    print("salary is not there")
```

```python
class Employee:
    language = "py"  #this is a class attribute
    salary = 100000
    
    def getInfo(self):
        print(f"The language is {self.language}  The salary is {self.salary}")
    
    @staticmethod    
    def greet():
        print("good morning")

        

        
ashish = Employee()
ashish.name = "Ashish" 
ashish.language = "javaScript" #this is a object/instance attribute



ashish.greet()
ashish.getInfo()
```

```python
# ANSWER NO -2.
# Write a class Calculator capable of finding square, 
# cube, and square root of a number.


class calculator:
    def __init__(self,n):
        self.n = n
        
    def square(self):
        print(f"The sqaure is {self.n*self.n}")
              
    def cube(self):
        print(f"The cube is {self.n*self.n*self.n}")
              
    def square_root(self):
        print(f"The square root is {self.n**(1/2)}")
                    
a = calculator(4)

a.square()
a.cube()
a.square_root()    
```

```python
# ANSWER NO - 4 .

# Add a static method in problem 2 to greet the user with "Hello".

# ANSWER NO -2.
# Write a class Calculator capable of finding square, 
# cube, and square root of a number.

# now i have to add an static method to greet user in this program


class calculator:
    def __init__(self,n):
        self.n = n
        
    def square(self):
        print(f"The sqaure is {self.n*self.n}")
              
    def cube(self):
        print(f"The cube is {self.n*self.n*self.n}")
              
    def square_root(self):
        print(f"The square root is {self.n**(1/2)}")
        
    @staticmethod
    def greet():
        print("Hello , Welcome !!")
                    
a = calculator(4)

calculator.greet()
a.square()
a.cube()
a.square_root()    


```

```python
# ANSWER NO - 5.

# Write a class Train which has methods to:

# Book a ticket
# Get status (number of seats)
# Get fare information of trains running under Indian Railways


class train:
    def __init__(self, name , seat ,fare ):
        self.name = name
        self.seat = seat
        self.fare = fare
        
    def get_status(self):
        print(f"The train {self.name} has {self.seat} seats available.")
    
    def get_fare_info(self):
        print(f"The fare of train {self.name }is {self.fare}")
        
    def book_ticket(self):
       
        if self.seat> 0 :
            print("Your ticket has been booked!!")
            self.seat -= 1
            print(f"Seats left : {self.seat}")
            
        else:
            print("sorry!! The train is full.")
            
t1 = train("Rajdhani_express" , 2 , 1200)

t1.get_status()
t1.get_fare_info()

t1.book_ticket()
t1.book_ticket()
t1.book_ticket()
t1.book_ticket()
t1.book_ticket()
```
