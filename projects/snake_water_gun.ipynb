User keeps guessing

Program tells:

> “Too high”

> “Too low”

> “Correct!”


Keep looping until correct guess

Count number of attempts

<b> Rules (important)

> Use while loop

> Use input()

> Use random

Handle invalid input (like letters)

```python
import random

computer = random.choice(range(1,101))

def guess():
    return int(input("Enter your guess between 1 to 100 :"))
attempts = 0  
a = guess()
attempts += 1

while(a != computer and attempts <7):
    
    if (a > computer):
        print("too high")
        
    elif (a<computer):
        print("too low")
   
    a = guess()
    attempts += 1

if (computer == a):
    print(f"You guessed it in {attempts} tries")
else: 
    print("game over the number was", computer)

    
```
