Build a Student Grade Manager in pure Python.

It should do the following:

1. Store 5 students as a dictionary where:
   - Key = student name (string)
   - Value = list of 3 marks (integers)

2. Write a function called get_average(marks) that:
   - Takes a list of marks
   - Returns the average (float, rounded to 2 decimal places)

3. Write a function called get_grade(average) that:
   - Returns "A" if average >= 85
   - Returns "B" if average >= 70
   - Returns "C" if average >= 55
   - Returns "D" if below 55

4. Loop through all students and print:
   "Ashish → Average: 82.33 → Grade: B"
   (formatted exactly like this using f-strings)

5. After printing all, print:
   - The name of the top scoring student
   - The class average across all students

```python
# STUDENT GRADE MANAGER IN PYTHON ~~

students = {"ashish" :[92,90,95],
            "ritik"  :[80,85,88],
            "vishaka":[75,85,78],
            "kunal"  :[88,65,58],
            "krish"  :[65,55,68]
           }
def get_average(marks):
    return sum(marks)/len(marks)

def get_grade(avg):
    if avg >= 85:
        return("A")
    elif avg >= 70:
        return("B")
    elif avg >=55:
        return("C")
    else:
        return("D")
    
for name,marks in students.items():
    avg = get_average(marks)
    grade = get_grade(avg)
    print(f"{name} → Average :{avg:.2f} → Grade:{grade}")
    

top_name=""
top_avg=0

for name,marks in students.items():
    avg =get_average(marks)
    if avg > top_avg:
        top_avg = avg
        top_name = name
        
# class average

total = 0

for marks in students.values():
    total += get_average(marks)
        
class_avg = total/len(students)

print(f"Top student: {top_name}")
print(f"class average: {class_avg:.2f}")
    
```
