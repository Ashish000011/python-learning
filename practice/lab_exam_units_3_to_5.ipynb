Read 5 numbers into a list. Apply `append(60)`, `insert(0, 5)`, `extend([70, 80])`, `remove(50)`, `pop()`, `sort()`, `reverse()`, `count(10)` and `index(20)`, printing the list with a label after each step.

**Sample:** `30 10 50 20 40` → `After append(60): [30, 10, 50, 20, 40, 60]`, then `After insert(0, 5): [5, 30, 10, 50, 20, 40, 60]`, and so on.

**3.4 · Lab level**

Print the elements greater than the average, and how many there are.

**Sample:** `10 20 30 40 50` → Average = 30.0, Greater: `[40, 50]` (2)

```python
def program_3_4():
    numbers = list(map(int, input("Enter numbers: ").split()))

    average = sum(numbers) / len(numbers)

    greater = [x for x in numbers if x > average]

    print("Average =", average)
    print("Greater:", greater)
    print("Count:", len(greater))


program_3_4()
```

**3.5 · Lab level**

Find the second smallest element without sorting. Print a message when there isn't one.

**Sample:** `7 3 9 3 5` → Second smallest = 5. `4 4 4` → No second smallest

```python
def program3_5():
    numbers = list(map(int,input("Enter the elements in list :").split()))

    smallest = None
    second_smallest = None

    for x in numbers:

        if smallest is None or x < smallest:
            second_smallest = smallest
            smallest = x
        elif x > smallest:
            if second_smallest is None or x < second_smallest:
                second_smallest = x

    if second_smallest is None:
        print("No second smallest")
    else:
        print("Second smallest =", second_smallest)

program3_5()
```

**3.7 · Lab level**

Check whether a list is in ascending order without using `sort()`.

**Sample:** `1 3 3 8` → Sorted. `1 5 2` → Not sorted

```python
def program3_7():
    numbers = list(map(int,input("Enter the elements of list seperated by space :").split()))

    sorted_list = True

    for i in range(len(numbers)-1):
        if numbers[i] > numbers[i+1]:
            sorted_list = False
            break
    if sorted_list:
        print("Sorted")
    else:
        print("Not sorted")

program3_7()    
```

**3.10 · Lab level**

A list holds the numbers 1 to n in any order, with one missing. Find the missing number.

**Sample:** `3 7 1 2 8 4 5` → Missing number = 6

```python
def program3_10():
    numbers = list(map(int, input("Enter elements: ").split()))

    numbers = sorted(numbers)
    count = 1

    for i in numbers:
        if i == count:
            count += 1
        else:
            print("Missing number:", count)
            return

    print("Missing number:", count)


program3_10()
```

**3.11 · Must-do · Harder**

Move all zeros to the front in place, keeping the order of the other numbers.

**Sample:** `0 1 0 3 12` → `[0, 0, 1, 3, 12]`

**3.12 · Harder**

Merge two sorted lists into one sorted list without `sort()` or `sorted()`.

**Sample:** `1 4 9` and `2 3 10 12` → `[1, 2, 3, 4, 9, 10, 12]`

```python
def program3_12():
    numbers = list(map(int, input("Enter list 1: ").split()))
    numbers1 = list(map(int, input("Enter list 2: ").split()))

    combined = numbers + numbers1
    result = []

    while len(combined) > 0:
        smallest = min(combined)
        result.append(smallest)
        combined.remove(smallest)

    print("Merged list:", result)


program3_12()
```

**3.13 · Harder**

Print the leaders: elements bigger than everything to their right. The last element is always a leader.

**Sample:** `16 17 4 3 5 2` → `[17, 5, 2]`

```python
def program3_13():
    numbers = list(map(int, input("Enter elements: ").split()))

    leaders = []

    for i in range(len(numbers)):
        flag = True

        for j in range(i+1, len(numbers)):
            if numbers[i] <= numbers[j]:
                flag = False
                break

        if flag == True:
            leaders.append(numbers[i])

    print("Leaders:", leaders)


program3_13()
```
