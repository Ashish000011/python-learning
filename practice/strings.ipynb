```python
# Q16. Take a sentence. Count:

# Number of vowels

# Number of consonants

# Number of spaces

# (All in one loop)

s = input("enter the sentence :")

vowels = ["a", "e" , "i" , "o" , "u"]

vowels_count = 0
consonants_count  = 0
space_count = 0

for char in s:
    if char in vowels:
        vowels_count += 1
    elif(char == " "):
        space_count += 1
    elif(char.isalpha()):
        consonants_count += 1


print(vowels_count)
print(consonants_count)
print(space_count)
```

```python
# Q17. Take a string. Print the frequency of each character (how many times each letter appears).

# Example: "hello" → h:1, e:1, l:2, o:1

s = input("Enter string: ")

seen = ""

for ch in s:
    if ch not in seen:
        print(ch, ":", s.count(ch))
        seen += ch
```

```python
# Q20. Take a sentence. Remove all vowels from it and print the result.

# Example: "Hello World" → "Hll Wrld"

s = input("Enter the sentence :")

vowels = ["a" , "e" , "i" , "o" , "u"]

new_s = []

for ch in s:
    if ch.lower() not in vowels:
        new_s.append(ch)
        
print("".join(new_s))
        
```

```python
# Q21. Take two strings from the user. Without using any built-in comparison method, check if the two strings are anagrams of each other.

# (Two words are anagrams if they have the exact same characters — e.g. "listen" and "silent")

s1 = input().lower()
s2 = input().lower()

if len(s1) != len(s2):
    print("Not anagram")
else:
    is_anagram = True

    for ch in s1:
        if s1.count(ch) != s2.count(ch):
            is_anagram = False
            break

    if is_anagram:
    
        print("Anagram")
    else:
        print("Not anagram")

    
```

```python
# Q22. Take a string. Find the longest word in it.

# Example: "I love programming" → "programming" (No sorting — use a loop!)

s = input("enter string: ")

s = s.split()

longest = [""]

for word in s:
    if(len(word) > len(longest)):
        longest = word

print(longest)
    
```

```python
# Q23. Build a Caesar Cipher encoder:

# Take a message and a shift number from the user

# Shift every letter forward by that number (a→b if shift=1, a→c if shift=2)

# Keep spaces as spaces, ignore non-letters

# Print the encoded message

# (Hint: use ord() and chr()!)

msg = input("enter msg :")
n = int(input("enter value to shift by :"))

result = ""

for ch in msg:
    if ch.isalpha():
        new_char = chr((ord(ch) - 97 + n) % 26 +97)
        result += new_char
    else:
        result += ch

print(result)

```

```python
# Q25. Take a paragraph (multi-word string) from the user. Print:

# Total word count

# The most repeated word (case-insensitive)

# All words that appear more than once

# (Loop-based only — no dictionaries!)

s = input("Write a paragraph :").lower()

words = s.split()
print(len(words))

max_count = 0
most_word = ""

for word in words:
    count = words.count(word)
    if count > max_count:
        max_count = count
        most_word = word

print("Most repeated word:", most_word)
printed = []

print("Repeated words:")
for word in words:
    if words.count(word) > 1 and word not in printed:
        print(word)
        printed.append(word)

    
        


```

```python
# Q26. Take a string. Compress it using Run Length Encoding:

# Consecutive repeated characters get replaced by character + count

# Example: "aaabbbccddddee" → "a3b3c2d4e2"

# If a character appears only once, just keep the character: "abc" → "abc"


s = "aaabbbccdddeeeaa"

result = ""
count = 1

for i in range(len(s)):
    if i < len(s) - 1 and s[i] == s[i+1]:
        count += 1
    else:
        result += s[i]

        if count > 1:
            result += str(count)

        count = 1

print(result)
        
    
```

```python
# Q29. Take a string. Check if it's a valid password by these rules:

# At least 8 characters long
# Contains at least one uppercase letter
# Contains at least one lowercase letter
# Contains at least one digit
# Contains at least one special character from !@#$%^&*
# Print exactly which rules are passing and which are failing

s = input("make password : ")

cases = ["!", "@", "#", "$", "%", "^", "&", "*"]


has_upper = False
has_lower = False
has_digit = False
has_special = False


if len(s) >= 8:
    print("Length condition passed")
else:
    print("Length condition failed")


for c in s:
    if c.isupper():
        has_upper = True
    if c.islower():
        has_lower = True
    if c.isdigit():
        has_digit = True
    if c in cases:
        has_special = True


print("Uppercase condition passed" if has_upper else "Uppercase condition failed")
print("Lowercase condition passed" if has_lower else "Lowercase condition failed")
print("Digit condition passed" if has_digit else "Digit condition failed")
print("Special character condition passed" if has_special else "Special character condition failed")

        
   
```

```python
# Q30. Take two strings. Find the longest common substring between them.

# Example: "ABCXYZ" and "XYZABC" → Longest common = "ABC" or "XYZ" (length 3) (Pure loop-based — no libraries)

s1 = "ABCXYZ"
s2 = "XYZABC"

longest = ""

for i in range(len(s1)):
    for j in range(i+1 , len(s1)+1):
        chunk = s1[i:j]
        if chunk in s2:
            if len(chunk)>len(longest):
                longest = chunk
print(longest)
    
```
