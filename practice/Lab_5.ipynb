## Q1. Print first 5 terms of Sine series
**Formula:** `sin(x) = x - x³/3! + x⁵/5! - x⁷/7! + x⁹/9!`

```python
import math
x = float(input("Enter n: "))

for i in range(5):
    power = 2*i+1
    term = x**power/math.factorial(power)

    if i % 2 == 1:
        term = -term
    print(term)
```

## Q6. Matrix Addition
Take two matrices and perform matrix addition.

⚠️ **No NumPy — write from scratch**

```python
r = int(input("Enter no. of rows: "))
c = int(input("Enter no. of columns: "))

A = []
B = []

# Input Matrix A
for i in range(r):
    row = []
    for j in range(c):
        x = int(input(f"Enter A[{i}][{j}]: "))
        row.append(x)
    A.append(row)

# Input Matrix B
for i in range(r):
    row = []
    for j in range(c):
        x = int(input(f"Enter B[{i}][{j}]: "))
        row.append(x)
    B.append(row)

# Addition
C = []

for i in range(r):
    row = []
    for j in range(c):
        row.append(A[i][j] + B[i][j])
    C.append(row)

# Display result
print("Result:")
for row in C:
    print(row)
```

## Q8. Transpose of a Matrix
Print the transpose of a given matrix.

⚠️ **No NumPy — write from scratch**

```python
r = int(input("Enter the no of rows in matrix : "))
c = int(input("Enter the no of columns in matrix : "))

matrix = []
for i in range(r):
    row = []
    for j in range(c):
        x = int(input(f"Enter matrix[{i}][{j}]: "))
        row.append(x)
    matrix.append(row)

print("Initial matrix :")
for i in matrix:
    print(i)
print()
# now transposing the matrix

transpose =[]

for i in  range(c):
    row = []
    for j in range(r):
        x = matrix[j][i]
        row.append(x)
    transpose.append(row)

print("Transposed matrix: ")
for i in transpose:
    print(i)
```
