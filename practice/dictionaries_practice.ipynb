- pythondata = {"a": 1, "b": 2, "c": 3, "d": 4, "e": 5}

```python
# Q8. Take this dictionary of student marks:

# pythonmarks = {"Aman": 85, "Priya": 92, "Rohan": 67, "Sneha": 45, "Karan": 78}
# Find and print the student with the highest marks (loop through .items() — no max()!)
# Find and print the student with the lowest marks

marks = {"Aman": 85, "Priya": 92, "Rohan": 67, "Sneha": 45, "Karan": 78}

highest_marks = float("-inf")
student_highest = ""
lowest_marks = float("inf")
student_lowest = ""
for key , value in marks.items():
    if(value > highest_marks):
        highest_marks = value
        student_highest = key
    if(value < lowest_marks):
        lowest_marks = value
        student_lowest = key
        

print("Student with highest marks is :\n",student_highest ,"->",highest_marks, sep="")
print("Student with lowest marks is :\n",student_lowest ,"->",lowest_marks, sep="")
```

```python
# Q9. Take a sentence from the user. Count the frequency of each word and store it in a dictionary.

# Input: "the cat sat on the mat the cat"
# Output: {"the": 3, "cat": 2, "sat": 1, "on": 1, "mat": 1}



data = input("enter sentence")

l = data.lower().split()
s = set(l)
dict = {i :l.count(i) for i in s } 
print(dict)


# order of o(n) solution

sentence = input("enter sentece")
words = sentence.lower().split()

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)
```
