```python
# Q19. Take two lists of numbers.

# Merge and sort them into one list without using any built-in sort on the combined list — sort it yourself using a loop.

list1 = [3, 7, 1, 9, 4]
list2 = [8, 2, 6, 5, 10]

m_list = list1 + list2
print("Merged list :", m_list)

s_list = []

while(len(m_list) > 0):
    largest = m_list[0]

    for i in m_list:
        if(i > largest):
            largest = i 

    s_list.insert(0,largest)
    m_list.remove(largest)
    
print("Sorted list :", s_list)

```

```python
# Q28. Take a list of integers.

# Find the longest consecutive sequence:
# [100, 4, 200, 1, 3, 2] → longest sequence is [1,2,3,4] → length 4
# (No sorting allowed! Pure loop logic)

list = [100, 4, 200, 1, 3, 2]
new_list = []
for i in list:
    for j in list:
        if(j == i+1) and (i) not in new_list:
            new_list.append(i)
            if (j not in new_list):
                new_list.append(j)

l_s_list = []



while(len(new_list) > 0):
    longest = new_list[0]
    for i in new_list:
        if(i > longest):
            longest = i
            
    l_s_list.insert(0,longest)
    new_list.remove(longest)
            
print(f"Longest sequence is {l_s_list} → Length {len(l_s_list)}")
        
        

```

```python
nums = [100, 4, 200, 1, 2, 3]

longest_seq = []
max_length = 0

for num in nums:
    if (num - 1) not in nums:
        current = num
        current_seq = [current]

        while (current + 1) in nums:
            current += 1
            current_seq.append(current)

        if len(current_seq) > max_length:
            max_length = len(current_seq)
            longest_seq = current_seq

print(f"Longest sequence is {longest_seq} -> Length {max_length}")
```
