```python
# Q17. Take a list of student marks:

# pythonmarks = [55, 82, 91, 37, 68, 44, 79, 23, 95, 61]

# Count:

# How many passed (>= 40)

# How many failed (< 40)

# Print the average marks of only passed students

marks = [55, 82, 91, 37, 68, 44, 79, 23, 95, 61]

count_passed = 0
count_failed = 0
sum_passed = 0

for i in marks:
    if i < 40:
        count_failed += 1
        continue      # ← skip the rest for failed students
    count_passed += 1
    sum_passed += i

avg_passed = sum_passed/count_passed

print(f"Total no of passed students : {count_passed}\nTotal no of failed students : {count_failed}\nThe average marks of passed students : {avg_passed}")

```

```python
# Q29. Build a student report card system:

# Ask how many students (say 5)
# For each student, ask their name + marks in 3 subjects
# Calculate each student's average
# Skip any student whose any subject mark is below 10 (data error) using continue
# Print a final table:

# Name        Avg     Grade
# -----------------------------
# Aman        82.3    B
# Priya       91.0    O
# ...



n = int(input("Enter number of students: "))

names = []
avgs = []
grades = []

for i in range(n):
    name = input("\nEnter student name: ")

    m1 = float(input("Enter marks in subject 1: "))
    m2 = float(input("Enter marks in subject 2: "))
    m3 = float(input("Enter marks in subject 3: "))

    # Skip invalid student
    if m1 < 10 or m2 < 10 or m3 < 10:
        print("Invalid marks! Skipping student...")
        continue

    avg = (m1 + m2 + m3) / 3

    # Grade calculation
    if avg >= 90:
        grade = "O"
    elif avg >= 75:
        grade = "A"
    elif avg >= 60:
        grade = "B"
    elif avg >= 50:
        grade = "C"
    else:
        grade = "F"

    # Store in lists
    names.append(name)
    avgs.append(avg)
    grades.append(grade)

# Final Table
print("\nName\tAvg\tGrade")
print("-----------------------------")

for i in range(len(names)):
    print(f"{names[i]}\t{avgs[i]:.1f}\t{grades[i]}")



## im accepting i used ai in this not fully but took a idea cause i got confused tht it can't be done without dictionary
```

```python
# Q30. Collatz Conjecture — one of the most famous math mysteries!

# Take any number from the user
# If even → divide by 2
# If odd → multiply by 3 and add 1
# Keep doing this until you reach 1
# Count and print how many steps it took
# Print every number in the sequence too

# (Example: 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1 = 8 steps)


n = int(input("Enter any number: "))

i = n   # initialize properly
count = 0

print(i, end=" -> ")

while i != 1:
    if i % 2 == 0:
        i = i // 2
    else:
        i = (i * 3) + 1

    count += 1
    print(i, end=" -> ")

print("\nSteps:", count)
```
