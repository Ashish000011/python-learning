```python
# Q20. Tuple Analysis (No Dictionary Allowed)

# Given the tuple:

# nums = (3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5)

# Find the following without using a dictionary (use loops and .count() only):

# The most repeated element
# All elements that appear more than once
# Total number of unique elements

nums = (3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5)

most_repeated = None
max_count = 0

for i in nums:
    count = nums.count(i)
    if count > max_count:
        max_count = count
        most_repeated = i

print("Most repeated element is :",most_repeated)
 
cheaked = []
unique_item = []

for i in nums :
    if i not in cheaked:
        unique_item.append(i)
        cheaked.append(i)

print("No. of unique elements are : ",len(unique_item))

more_then_once = []
for i in nums:
    if(nums.count(i) > 1):
        if( i not in more_then_once):
            more_then_once.append(i)

print("Elements present more then once are : ",more_then_once)
```

```python
# Q23. You have a tuple of monthly sales figures:
# pythonsales = (12000, 15000, 9000, 18000, 21000, 17000, 14000, 22000, 19000, 11000, 16000, 20000)
# Calculate and print:
# Best month (highest sales) and its month name
# Worst month (lowest sales) and its month name
# Total annual sales
# Monthly average
# How many months exceeded the average
# The trend: was the second half of the year better than the first? Print "Better 📈" or "Worse 📉"
# (months = ("Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"))

sales = (12000, 15000, 9000, 18000, 21000, 17000, 14000, 22000, 19000, 11000, 16000, 20000)
months = ("Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec")

best_month = months[0]
worst_month = months[0]

highest_sales = sales[0]
lowest_sales = sales[0]

for i, value in enumerate(sales):
    if(value > highest_sales):
        highest_sales = value
        best_month = months[i]

    if(value < lowest_sales):
        lowest_sales = value
        worst_month = months[i]

print("Best month:", best_month, highest_sales)
print("Worst month:", worst_month, lowest_sales)

total_annual_sales = 0

for i in sales:
    total_annual_sales += i 

print("Total Annual sales :",total_annual_sales)
monthly_avg = total_annual_sales//len(sales)

print("Monthly Average :", monthly_avg)

months_excedding = []

for i, value in enumerate(sales):
    if(value > monthly_avg):
        months_excedding.append(months[i])

print("Months which exceeded monthly avg :", " , ".join(months_excedding))

mid = len(sales)//2
first_half = sales[:mid]
second_half = sales[mid:]

total_first_half = 0
total_second_half = 0

for i in first_half:
    total_first_half += i 
for i in second_half:
    total_second_half += i 

if(total_second_half> total_first_half):
    print("Trend : Better 📈")
else:
    print("Trend : Worse 📉")
    

```

```python
# Q24. Take two tuples from the user (5 numbers each, entered in a loop). Without converting to lists or using any sort:

# Find common elements between them
# Find elements only in tuple 1
# Find elements only in tuple 2
# Merge and print them sorted (build sorted result manually)
# Take input for tuple 1


tuple1 = ()
for i in range(5):
    a = int(input("Enter element of tuple 1: "))
    tuple1 = tuple1 + (a,)

print("NOW FOR TUPLE 2")
tuple2 = ()
for i in range(5):
    a = int(input("Enter element of tuple 2: "))
    tuple2 = tuple2 + (a,)

common_elements = ()
for i in tuple1:
    if i in tuple2:
        common_elements = common_elements + (i,)

print("Common elements in both tuples:", common_elements)

only_tuple1 = ()
for i in tuple1:
    if i not in common_elements:
        only_tuple1 = only_tuple1 + (i,)

print("Elements only in tuple 1:", only_tuple1)

only_tuple2 = ()
for i in tuple2:
    if i not in common_elements:
        only_tuple2 = only_tuple2 + (i,)

print("Elements only in tuple 2:", only_tuple2)


merged = only_tuple1 + only_tuple2


merged_sorted = ()

temp = merged 

while len(temp) > 0:
    smallest = temp[0]

    for i in temp:
        if i < smallest:
            smallest = i

    merged_sorted = merged_sorted + (smallest,)
    
    new_temp = ()
    removed = False
    for i in temp:
        if i == smallest and not removed:
            removed = True
        else:
            new_temp = new_temp + (i,)
    
    temp = new_temp

print("Sorted merged elements:", merged_sorted)
```
